```python

from utils.config_loader import load_env
import pandas as pd
from src.llm_agent import get_agent
from utils.helper import normalize_rows
```

```python
generate_sql, db = get_agent()

question = "how many orders in last 1 year?"
sql = generate_sql(question)

print("Generated SQL:")
print(sql)
print("-" * 80)

try:
    raw_rows = db.run(sql)
    print("Raw DB output:", raw_rows)
except Exception as e:
    print("❌ Query failed:", e)
else:
    result = normalize_rows(raw_rows)

    print("\nNormalized Result:")
    if result is None:
        print("(No rows returned)")

    elif isinstance(result, pd.DataFrame):
        print("DataFrame with shape:", result.shape)
        print(result)           # Jupyter will show DataFrame nicely
    else:
        print("Scalar value:")
        print(result)           # Scalar value
```
